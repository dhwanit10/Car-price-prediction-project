```python
import pandas as pd
import numpy as np
import json
```

```python
df = pd.read_csv("../data/processed/cardekho_enhanced.csv")

# print(df.head())

# brands = df["brand"].drop_duplicates().to_list()
brands = sorted(df["brand"].dropna().unique().tolist())

with open("../app/data/brands.json", "w", encoding="utf-8") as f:
    json.dump(brands, f, indent=4, ensure_ascii=False)
```

```python
models_by_brand = (
    df.groupby("brand")["model"]
      .apply(lambda x: sorted(x.dropna().unique().tolist()))
      .to_dict()
)

with open("../app/data/models.json", "w", encoding="utf-8") as f:
    json.dump(models_by_brand, f, indent=4, ensure_ascii=False)
```
